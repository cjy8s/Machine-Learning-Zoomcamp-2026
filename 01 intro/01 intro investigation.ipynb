```python
hp_median = cars['horsepower'].median()
hp_mode = cars['horsepower'].mode()[0]

print('median:', hp_median)
print('most frequent:', hp_mode)
hp_filled = cars['horsepower'].fillna(hp_mode)
hp_filled_median = hp_filled.median()
print('median:', hp_filled_median)

print('Has it changed?', hp_filled_median != hp_median)
```

```python
asian_cars = cars[cars['origin'] == 'Asia']
X = asian_cars[['vehicle_weight', 'model_year']].head(7).to_numpy()

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.linalg.inv(X.T @ X) @ X.T @ y
sum(w)
```
